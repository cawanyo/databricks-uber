```python
from  pyspark.sql.functions import *
from pyspark.sql.types import *

```

```python
for file in dbutils.fs.ls("abfss://uber-data@storageuberdb.dfs.core.windows.net/data"):

    if file.name == "bulk_rides.json" and spark.catalog.tableExists("uber.dev.bulk_rides"):
        continue
    
    df = spark.read \
                .option("multiLine", "true") \
                .option("mode", "PERMISSIVE") \
                .json(file.path)
    
    df.write.format('delta').mode("overwrite").saveAsTable("uber.dev." + file.name[0:-5])
```
